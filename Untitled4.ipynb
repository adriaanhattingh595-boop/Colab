### Persisting Configurations and Files to Google Drive

Google Colab environments are ephemeral. To preserve your custom application settings, SSH configurations, and files across session resets, we will use your Google Drive as a persistent storage layer.

First, we mount Google Drive, and then we will create helper scripts to **Snapshot** (save) and **Restore** (load) your settings.

````python
from google.colab import drive
import os
import shutil

# 1. Mount Google Drive
drive.mount('/content/drive')

# 2. Define our persistence directory on Google Drive
PERSIST_DIR = '/content/drive/MyDrive/ColabPersistence'
os.makedirs(PERSIST_DIR, exist_ok=True)
print(f'Persistent backup directory is ready at: {PERSIST_DIR}')
````

#### Create Backup and Restore Utilities

We will define paths we want to persist, such as:
- The SSH Configuration (`/etc/ssh/sshd_config`)
- VNC Configuration (`~/.vnc`)
- Your home directory workspace files

````python
import tarfile

# List of directories/files to back up
PATHS_TO_PERSIST = [
    '/etc/ssh/sshd_config',
    os.path.expanduser('~/.vnc'),
    '/root/.ssh'
]

def take_snapshot():
    """Archives specified configuration directories and saves them to Google Drive."""
    archive_path = os.path.join(PERSIST_DIR, 'colab_config_snapshot.tar.gz')
    print("Creating snapshot of configurations on Google Drive...")

    with tarfile.open(archive_path, "w:gz") as tar:
        for path in PATHS_TO_PERSIST:
            if os.path.exists(path):
                tar.add(path, arcname=os.path.relpath(path, '/'))
                print(f"Backed up: {path}")
            else:
                print(f"Skipped (does not exist yet): {path}")

    print(f"Snapshot successfully saved to {archive_path}")

def restore_snapshot():
    """Extracts the saved snapshot from Google Drive back to the root file system."""
    archive_path = os.path.join(PERSIST_DIR, 'colab_config_snapshot.tar.gz')
    if not os.path.exists(archive_path):
        print("No snapshot found on Google Drive to restore.")
        return

    print("Restoring configurations from Google Drive...")
    with tarfile.open(archive_path, "r:gz") as tar:
        tar.extractall(path='/')
    print("Configurations successfully restored!")
````

### 1. Headless Inference Server Setup (FastAPI & Uvicorn)

We will install FastAPI, Uvicorn, and a simple model runner to host an OpenAI-compatible endpoint. This API will listen on all interfaces so you can query it via your Tailscale IP (`http://100.70.83.83:8000/v1`).

````python
# Install FastAPI, Uvicorn, and HuggingFace Transformers for inference
!pip install -q fastapi uvicorn transformers pydantic
````

````python
import uuid

# Generate a static API Key for securing your local connection
API_KEY = f"colab-sk-{uuid.uuid4().hex[:12]}"
print(f"--- Remote AI Inference Credentials ---")
print(f"Base URL: http://100.70.83.83:8000/v1")
print(f"API Key:  {API_KEY}")
````

````python
import threading
import uvicorn
import time
from fastapi import FastAPI, Depends, HTTPException, Security
from fastapi.security.api_key import APIKeyHeader
from pydantic import BaseModel
from transformers import pipeline

app = FastAPI(title="Colab Remote Inference Engine")
API_KEY_HEADER = APIKeyHeader(name="Authorization", auto_error=True)

def get_api_key(api_key_header: str = Security(API_KEY_HEADER)):
    token = api_key_header.replace("Bearer ", "").strip()
    if token == API_KEY:
        return token
    raise HTTPException(status_code=403, detail="Could not validate credentials")

# Set up activity tracking variables for the idle daemon
LAST_ACTIVITY_TIME = time.time()

@app.middleware("http")
async def record_activity(request, call_next):
    global LAST_ACTIVITY_TIME
    LAST_ACTIVITY_TIME = time.time()
    return await call_next(request)

# Load a small generative model (GPT-2) as a headless test
print("Loading lightweight Hugging Face model...")
generator = pipeline('text-generation', model='gpt2', device=0)

class CompletionRequest(BaseModel):
    prompt: str
    max_tokens: int = 50

@app.post("/v1/completions")
def completions(request: CompletionRequest, token: str = Depends(get_api_key)):
    result = generator(request.prompt, max_new_tokens=request.max_tokens, num_return_sequences=1)
    return {
        "choices": [
            {
                "text": result[0]["generated_text"][len(request.prompt):],
                "index": 0,
                "finish_reason": "length"
            }
        ]
    }

# Run server in a background thread to keep Colab interface responsive
def run_server():
    uvicorn.run(app, host="0.0.0.0", port=8000, log_level="info")

threading.Thread(target=run_server, daemon=True).start()
print("Headless API Server (with activity tracking) successfully running in background on port 8000!")
````

### 2. Auto Incremental Backup to Google Drive

This background script runs a continuous loop that checks for modified files in `/content` and backs them up along with this notebook automatically every 15 minutes.

````python
import time
import subprocess

def background_backup_loop():
    while True:
        try:
            print("--- Triggering Auto Incremental Backup ---")
            # Save current configurations
            take_snapshot()
            # Safely attempt to copy the current Colab Notebook state over to Google Drive
            !cp -r /content/*.ipynb {PERSIST_DIR}/ 2>/dev/null || true
            print("Incremental backup sync complete.")
        except Exception as e:
            print(f"Backup error: {e}")
        time.sleep(900) # Backup every 15 minutes

threading.Thread(target=background_backup_loop, daemon=True).start()
````

### 3. Idle Auto-Sleep (Preserve Compute Units)

This function monitors active model API requests. If no requests are received for 30 minutes, it will automatically shut down the Colab runtime to prevent unwanted credit usage.

````python
from google.colab import runtime
import time
import threading

IDLE_TIMEOUT_SECONDS = 1800 # 30 minutes of idle

def idle_monitor():
    global LAST_ACTIVITY_TIME
    while True:
        time.sleep(60)
        elapsed = time.time() - LAST_ACTIVITY_TIME
        if elapsed > IDLE_TIMEOUT_SECONDS:
            print("System idle for over 30 minutes. Shutting down runtime to preserve compute credits!")
            runtime.unassign()

threading.Thread(target=idle_monitor, daemon=True).start()
print("Auto-sleep idle safety daemon activated.")
````

````python
import subprocess
import os
import time

def verify_and_start_services():
    print("=== Nexus Remote Suite Self-Healing & Startup Verification ===\n")

    # 1. Verify Tailscale Daemon
    ts_check = subprocess.getoutput("pgrep tailscaled")
    if not ts_check:
        print("[!] Tailscale daemon is down. Starting...")
        os.system("nohup sudo /usr/local/bin/tailscaled --tun=userspace-networking > tailscaled_daemon.log 2>&1 &")
        time.sleep(2)
    else:
        print(f"[✔] Tailscale Daemon is active (PID: {ts_check})")

    # 2. Verify SSH Service
    ssh_check = subprocess.getoutput("pgrep sshd")
    if not ssh_check:
        print("[!] SSH Service is down. Starting...")
        os.system("sudo service ssh start")
    else:
        print(f"[✔] OpenSSH Server is active (PID: {ssh_check})")

    # 3. Verify VNC Server
    vnc_check = subprocess.getoutput("pgrep Xvnc")
    if not vnc_check:
        print("[!] VNC Server is down. Starting...")
        os.system("vncserver -kill :1 2>/dev/null || true")
        os.system("vncserver :1 -geometry 1280x800 -depth 24")
    else:
        print(f"[✔] VNC Server (Xvnc) is active (PID: {vnc_check})")

    # 4. Verify noVNC Web Client
    novnc_check = subprocess.getoutput("pgrep -f novnc")
    if not novnc_check:
        print("[!] noVNC service is down. Starting...")
        os.system("nohup sudo /usr/share/novnc/utils/launch.sh --vnc localhost:5901 --listen 8080 > novnc.log 2>&1 &")
    else:
        print(f"[✔] noVNC Web Portal is active (PID: {novnc_check})")

    # 5. Verify FastAPI Headless Inference Server
    fastapi_check = subprocess.getoutput("pgrep -f uvicorn")
    if not fastapi_check:
        print("[!] FastAPI Headless Server is down. Starting server...")
        # Import to trigger background thread if not already running
        import threading
        try:
            threading.Thread(target=run_server, daemon=True).start()
        except NameError:
            print("FastAPI thread helper missing in global namespace; please execute the FastAPI code cell to bind standard imports.")
    else:
        print(f"[✔] FastAPI Inference API is active (PID: {fastapi_check})")

    # 6. Verify Streamlit Web GUI
    streamlit_check = subprocess.getoutput("pgrep -f 'streamlit run'")
    if not streamlit_check:
        print("[!] Streamlit GUI application is down. Starting...")
        os.system("nohup streamlit run app.py --server.port 8501 --server.address 0.0.0.0 > streamlit.log 2>&1 &")
    else:
        print(f"[✔] Streamlit GUI Application is active (PID: {streamlit_check})")

verify_and_start_services()
````

````python
import tarfile
import os
from datetime import datetime

def create_dated_full_backup():
    # Generate timestamp
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    backup_filename = f"nexus_suite_backup_{timestamp}.tar.gz"
    backup_path = os.path.join(PERSIST_DIR, backup_filename)

    print(f"Initializing complete system backup at: {timestamp}")

    # Paths to back up
    files_to_backup = [
        "/content/app.py",
        "/content/nexus_ai_suite.db",
        "/etc/ssh/sshd_config",
        os.path.expanduser("~/.vnc"),
        "/root/.ssh"
    ]

    # Create a compressed tar archive of our configurations and database
    with tarfile.open(backup_path, "w:gz") as tar:
        for path in files_to_backup:
            if os.path.exists(path):
                tar.add(path, arcname=os.path.relpath(path, "/"))
                print(f"Archived: {path}")
            else:
                print(f"Skipping {path} (not created yet)")

        # Safely capture any notebook files in /content
        for file in os.listdir("/content"):
            if file.endswith(".ipynb"):
                notebook_path = os.path.join("/content", file)
                tar.add(notebook_path, arcname=f"content/{file}")
                print(f"Archived Notebook: {file}")

    print(f"\n[✔] Complete dated backup file created successfully at:\n{backup_path}")
    print("This backup can be downloaded or synced directly from your Google Drive/OneDrive folder!")

create_dated_full_backup()
````
